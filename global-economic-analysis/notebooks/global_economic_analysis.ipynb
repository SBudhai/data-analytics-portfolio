```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("/content/Global Economy Indicators.csv")

print("Dataset loaded!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
print("COLUMN NAMES:")
for col in df.columns:
    print(col)

print("\nMISSING VALUES:")
print(df.isnull().sum())

print("\nDUPLICATE ROWS:")
print(df.duplicated().sum())

print("\nDATA TYPES:")
print(df.dtypes)
```

```python
# Remove leading/trailing spaces
df.columns = df.columns.str.strip()

# Rename columns we plan to use
df = df.rename(columns={
    "CountryID": "Country_ID",
    "AMA exchange rate": "AMA_Exchange_Rate",
    "IMF based exchange rate": "IMF_Exchange_Rate",
    "Per capita GNI": "GNI_Per_Capita",
    "Gross National Income(GNI) in USD": "GNI_USD",
    "Gross Domestic Product (GDP)": "GDP",
    "Exports of goods and services": "Exports",
    "Imports of goods and services": "Imports",
    "Final consumption expenditure": "Final_Consumption",
    "General government final consumption expenditure": "Government_Consumption",
    "Household consumption expenditure (including Non-profit institutions serving households)": "Household_Consumption",
    "Gross capital formation": "Gross_Capital_Formation",
    "Manufacturing (ISIC D)": "Manufacturing",
    "Agriculture, hunting, forestry, fishing (ISIC A-B)": "Agriculture"
})

print("Columns cleaned!")
print(df.columns.tolist())
```

```python
# GDP per capita
df["GDP_Per_Capita"] = df["GDP"] / df["Population"]

# Trade balance
df["Trade_Balance"] = df["Exports"] - df["Imports"]

# Export and import shares of GDP
df["Exports_Pct_GDP"] = (df["Exports"] / df["GDP"]) * 100
df["Imports_Pct_GDP"] = (df["Imports"] / df["GDP"]) * 100

# GDP growth by country
df = df.sort_values(["Country", "Year"])

df["GDP_Growth_Pct"] = (
    df.groupby("Country")["GDP"]
      .pct_change(fill_method=None) * 100
)

print("Financial indicators created!")

df[
    [
        "Country",
        "Year",
        "GDP",
        "GDP_Per_Capita",
        "Exports",
        "Imports",
        "Trade_Balance",
        "GDP_Growth_Pct"
    ]
].head(10)
```

```python
print("YEAR RANGE")
print(df["Year"].min(), "-", df["Year"].max())

print("\nNUMBER OF COUNTRIES/AREAS")
print(df["Country"].nunique())

print("\nNUMBER OF RECORDS")
print(len(df))

print("\nMOST RECENT YEAR")
print(df["Year"].max())

print("\nSAMPLE COUNTRIES")
print(sorted(df["Country"].unique())[:30])
```

```python
latest_year = df["Year"].max()

latest = df[df["Year"] == latest_year].copy()

print("Latest year:", latest_year)
print("Countries/areas in latest year:", latest["Country"].nunique())

latest[
    [
        "Country",
        "GDP",
        "GDP_Per_Capita",
        "GNI_USD",
        "Exports",
        "Imports",
        "Trade_Balance"
    ]
].sort_values("GDP", ascending=False).head(20)
```

```python
# Look for possible regional/world aggregates
possible_aggregates = [
    "World",
    "Europe",
    "Asia",
    "Africa",
    "Americas",
    "North America",
    "South America",
    "European Union"
]

df[
    df["Country"].str.strip().isin(possible_aggregates)
]["Country"].unique()
```

```python
# GDP in billions of USD
df["GDP_Billions"] = df["GDP"] / 1_000_000_000

# GNI in billions of USD
df["GNI_Billions"] = df["GNI_USD"] / 1_000_000_000

# Exports and imports in billions
df["Exports_Billions"] = df["Exports"] / 1_000_000_000
df["Imports_Billions"] = df["Imports"] / 1_000_000_000

# Trade balance in billions
df["Trade_Balance_Billions"] = df["Trade_Balance"] / 1_000_000_000

# Population in millions
df["Population_Millions"] = df["Population"] / 1_000_000

print("Dashboard fields created!")

df[
    [
        "Country",
        "Year",
        "GDP_Billions",
        "GDP_Per_Capita",
        "GDP_Growth_Pct",
        "Trade_Balance_Billions",
        "Population_Millions"
    ]
].tail()
```

```python
latest_analysis = df[
    (df["Year"] == 2021) &
    (df["Population"] >= 1_000_000)
].copy()

latest_analysis[
    [
        "Country",
        "GDP_Billions",
        "GDP_Per_Capita",
        "GDP_Growth_Pct",
        "Trade_Balance_Billions"
    ]
].sort_values(
    "GDP_Growth_Pct",
    ascending=False
).head(15)
```

```python
# Select fields for the Tableau dashboard
tableau_df = df[
    [
        "Country",
        "Year",
        "Population",
        "Population_Millions",
        "GDP",
        "GDP_Billions",
        "GDP_Per_Capita",
        "GDP_Growth_Pct",
        "GNI_USD",
        "GNI_Billions",
        "Exports",
        "Exports_Billions",
        "Imports",
        "Imports_Billions",
        "Trade_Balance",
        "Trade_Balance_Billions"
    ]
].copy()

# Rename columns for cleaner Tableau field names
tableau_df = tableau_df.rename(columns={
    "GDP_Growth_Pct": "GDP_Change_Pct"
})

print("Rows:", len(tableau_df))
print("Columns:", len(tableau_df.columns))

tableau_df.head()
```

```python
print("FINAL DATASET CHECK")
print("-------------------")

print("Rows:", len(tableau_df))
print("Countries:", tableau_df["Country"].nunique())
print(
    "Years:",
    tableau_df["Year"].min(),
    "-",
    tableau_df["Year"].max()
)

print("\nMissing values:")
print(tableau_df.isnull().sum())
```

```python
file_name = "global_economy_cleaned.csv"

tableau_df.to_csv(file_name, index=False)

print(f"Saved: {file_name}")
```
